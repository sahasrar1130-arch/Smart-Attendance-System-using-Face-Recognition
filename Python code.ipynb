```python
import cv2
import os
# Set the student name
student_name = "Student1"
# Create the folder path for this student
save_path = os.path.join(
    r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\student_images",
    student_name
)
# Create the folder if it doesn't already exist
os.makedirs(save_path, exist_ok=True)
# Open the default webcam
camera = cv2.VideoCapture(0)
# Start the image counter
count = 0
# Keep the camera running
while True:
    # Capture one frame from the webcam
    ret, frame = camera.read()
# Check if the camera returned a frame
    if not ret:
        print("Cannot access camera")
        break
         # Display the live camera feed

    cv2.imshow("Capture Face - Press Q to Stop", frame)
# Read a keyboard key
    key = cv2.waitKey(1) & 0xFF
# Press S to save the current frame

    if key == ord("s"):
        # Create a filename such as 0.jpg, 1.jpg, 2.jpg
        filename = os.path.join(save_path, f"{count}.jpg")
        # Save the captured image
        cv2.imwrite(filename, frame)
        # Increase the image counter
        count += 1
        # Display the saved image path
        print("Saved:", filename)
# Press Q to stop capturing
    elif key == ord("q"):
        break
# Release the webcam
camera.release()
# Close all OpenCV windows
cv2.destroyAllWindows()
# Display the total number of saved images

print("Total images saved:", count)
```

```python
import cv2
import os
# Path containing the original student images
student_path = r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\student_images\Student1"
# Path where detected face images will be saved
training_path = r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\training_data\Student1"
# Create the training folder if it does not exist
os.makedirs(training_path, exist_ok=True)
# Get the path of the Haar Cascade face detector
cascade_path = os.path.join(
    cv2.data.haarcascades,
    "haarcascade_frontalface_default.xml"
)
# Load the face detector
face_cascade = cv2.CascadeClassifier(cascade_path)
# Counter for saved training images
saved = 0
# Go through all files in Student1's image folder
for image_name in os.listdir(student_path):
# Process only image files
    if not image_name.lower().endswith((".jpg", ".jpeg", ".png")):
        continue
 # Create the complete image path
    image_path = os.path.join(student_path, image_name)
     # Read the image
    image = cv2.imread(image_path)
# Check whether the image was loaded
    if image is None:
        continue
 # Convert the image to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
# Detect faces in the image
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    # Save only images where exactly one face is detected
    if len(faces) == 1:
        # Get the face coordinates
        x, y, w, h = faces[0]
 # Crop only the detected face
        face = gray[y:y+h, x:x+w]
 # Create a filename for the cropped face
        filename = os.path.join(
            training_path,
            f"Student1_{saved}.jpg"
        )
# Save the cropped face
        cv2.imwrite(filename, face)
        # Increase the counter
        saved += 1

        print("Saved:", filename)

    else:
        # Skip images with zero or multiple faces
        print("Skipped:", image_name, "Faces detected:", len(faces))
# Display the total number of training images
print("\nTotal training images:", saved)
```

```python
import cv2
import os

# Path where the training images are stored
training_path = r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\training_data"
# Lists to store face images and their corresponding labels
faces = []
labels = []

# Assign a numeric label to the student
student_label = 0
# Name of the student folder
student_name = "Student1"
# Create the complete path to the student's folder
student_folder = os.path.join(training_path, student_name)
# Read every file inside the student's folder
for image_name in os.listdir(student_folder):
 # Process only image files
    if not image_name.lower().endswith((".jpg", ".jpeg", ".png")):
        continue
    # Create the complete path of the image
    image_path = os.path.join(student_folder, image_name)
 # Read the image in grayscale
    # Grayscale is required for the LBPH face recognizer
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
 # Check whether the image was successfully loaded
    if image is not None:
            # Add the image to the training images list
        faces.append(image)
         # Add the student's numeric label
        labels.append(student_label)
# Display the number of images used for training
print("Training images:", len(faces))

# Display the labels assigned to the images
print("Labels:", labels)

# Create the LBPH (Local Binary Patterns Histograms) face recognizer
recognizer = cv2.face.LBPHFaceRecognizer_create()

# Convert labels into a NumPy array and train the recognizer
recognizer.train(faces, __import__("numpy").array(labels))

# Path where the trained face recognition model will be saved
model_path = r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\student_model.yml"

# Save the trained model as a YAML file
recognizer.write(model_path)

# Display training completion message
print("Training completed!")

# Display the location of the saved model
print("Model saved at:", model_path)
```

```python
import cv2

model_path = r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\student_model.yml"

recognizer = cv2.face.LBPHFaceRecognizer_create()
recognizer.read(model_path)

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

camera = cv2.VideoCapture(0)

while True:
    ret, frame = camera.read()

    if not ret:
        print("Camera error")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    for (x, y, w, h) in faces:

        face = gray[y:y+h, x:x+w]

        label, confidence = recognizer.predict(face)

        if confidence < 70:
            name = "Student1"
        else:
            name = "Unknown"

        cv2.rectangle(
            frame,
            (x, y),
            (x+w, y+h),
            (0, 255, 0),
            2
        )

        cv2.putText(
            frame,
            name,
            (x, y-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 0),
            2
        )

        cv2.putText(
            frame,
            f"Confidence: {confidence:.2f}",
            (x, y+h+25),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )

    cv2.imshow("Smart Attendance - Face Recognition", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

camera.release()
cv2.destroyAllWindows()
```

```python
import cv2
import csv
import os
from datetime import datetime


# 1. FILE PATHS

# Path of the trained face recognition model
model_path = r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\student_model.yml"

# New attendance file
# Using a new file avoids the PermissionError from attendance.csv
attendance_file = r"C:\Users\SAHASRA\Desktop\SMART ATTENDENCE SYSTEM\attendance_new.csv"


# 2. CHECK OPENCV FACE MODULE

if not hasattr(cv2, "face"):
    print("ERROR: cv2.face is not available!")
    print("Install OpenCV Contrib using:")
    print("pip uninstall opencv-python opencv-contrib-python -y")
    print("pip install opencv-contrib-python==4.10.0.84")
    raise SystemExit


# 3. CHECK MODEL FILE


if not os.path.exists(model_path):
    print("ERROR: Model file not found!")
    print("Expected file:")
    print(model_path)
    raise SystemExit

print("Model file found!")


# 4. LOAD FACE RECOGNIZER

recognizer = cv2.face.LBPHFaceRecognizer_create()

recognizer.read(model_path)

print("Face recognition model loaded successfully!")


# 5. LOAD HAAR CASCADE FACE DETECTOR
cascade_path = cv2.data.haarcascades + \
               "haarcascade_frontalface_default.xml"

face_cascade = cv2.CascadeClassifier(cascade_path)

if face_cascade.empty():
    print("ERROR: Face detector could not be loaded!")
    raise SystemExit

print("Face detector loaded successfully!")

# 6. CREATE ATTENDANCE FILE

if not os.path.exists(attendance_file):

    try:
        with open(
            attendance_file,
            "w",
            newline=""
        ) as file:

            writer = csv.writer(file)

            # Create column headings
            writer.writerow([
                "Name",
                "Date",
                "Time",
                "Status"
            ])

        print("Attendance file created!")

    except PermissionError:
        print("ERROR: Permission denied while creating attendance file.")
        print("Close Excel or any program using the CSV file.")
        raise SystemExit

# 7. ATTENDANCE CONTROL

# Prevent multiple attendance entries
# during the same camera session
marked = False

# 8. OPEN CAMERA

camera = cv2.VideoCapture(0)

if not camera.isOpened():
    print("ERROR: Camera could not be opened!")
    print("Try changing:")
    print("camera = cv2.VideoCapture(0)")
    print("to:")
    print("camera = cv2.VideoCapture(1)")
    raise SystemExit

print("Camera started successfully!")
print("Look at the camera.")
print("Press Q to quit.")


# 9. START CAMERA LOOP

while True:

    # Capture a frame
    ret, frame = camera.read()

    if not ret:
        print("ERROR: Could not read camera frame.")
        break

    # 10. CONVERT FRAME TO GRAYSCALE


    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )


    # 11. DETECT FACES
  
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )


    # 12. PROCESS EACH FACE
 
    for (x, y, w, h) in faces:

        # Extract detected face
        face = gray[y:y+h, x:x+w]

        # 13. RECOGNIZE FACE
  

        label, confidence = recognizer.predict(face)

        # 14. CHECK STUDENT
  
        # Student1 was trained with label 0
        # Lower LBPH distance generally means a better match
        if label == 0 and confidence < 70:

            name = "Student1"

            # 15. MARK ATTENDANCE
            if not marked:

                now = datetime.now()

                try:

                    # Open CSV file in append mode
                    with open(
                        attendance_file,
                        "a",
                        newline=""
                    ) as file:

                        writer = csv.writer(file)

                        # Add attendance record
                        writer.writerow([
                            name,
                            now.strftime("%Y-%m-%d"),
                            now.strftime("%H:%M:%S"),
                            "Present"
                        ])

                    # Attendance successfully saved
                    marked = True

                    print()
                    print("--------------------------------")
                    print("ATTENDANCE MARKED SUCCESSFULLY")
                    print("--------------------------------")
                    print("Name :", name)
                    print("Date :", now.strftime("%Y-%m-%d"))
                    print("Time :", now.strftime("%H:%M:%S"))
                    print("File :", attendance_file)
                    print("--------------------------------")
                    print()

                except PermissionError:

                    print()
                    print("--------------------------------")
                    print("ERROR: Permission denied!")
                    print("--------------------------------")
                    print("Please close attendance_new.csv")
                    print("if it is open in Excel.")
                    print("--------------------------------")
                    print()


        else:

            # Face does not match Student1
            name = "Unknown"

        # 16. DRAW FACE RECTANGLE

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )


        # 17. DISPLAY NAME
        cv2.putText(
            frame,
            name,
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 0),
            2
        )

        # 18. DISPLAY DISTANCE
    

        cv2.putText(
            frame,
            f"Distance: {confidence:.2f}",
            (x, y + h + 25),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )

    # 19. DISPLAY CAMERA WINDOW

    cv2.imshow(
        "Smart Attendance System",
        frame
    )

    # 20. PRESS Q TO EXIT
    
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break



# 21. RELEASE CAMERA

camera.release()


# 22. CLOSE WINDOWS

cv2.destroyAllWindows()



# 23. FINAL MESSAGE


print()
print("========================================")
print("Smart Attendance System Stopped")
print("========================================")
print("Attendance file:")
print(attendance_file)
print("========================================")
```
